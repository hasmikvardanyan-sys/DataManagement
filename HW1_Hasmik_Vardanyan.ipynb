### 1.3 Boolean masks and replacement — 8 points

Extract all values in `units` that are at least 15 and less than 30 using a Boolean mask. Then use `np.where` to create `adjusted_units`, replacing values below 10 with 0 and keeping all other values. Display `adjusted_units - units`. Explain why `&` and parentheses are needed when combining the two array conditions.

<!-- Solution hint: define lower_edge = 3 * 5; upper_edge = lower_edge + lower_edge; cutoff = upper_edge // 3; then use (units >= lower_edge) & (units < upper_edge) for selection and np.where(units < cutoff, 0, units) for replacement. -->

### 1.4 Summary calculations — 10 points

Using the original `units`, calculate: total units for each day; mean daily units for each product; median units across the four products for each day; and the overall population standard deviation (use `np.std` with its default settings). Label which result belongs to which day or product. Explain what `axis=0` and `axis=1` mean for this array.

<!-- Solution hint: define swapped_units = units.T; compute daily totals with np.sum(swapped_units, axis=0), product means with np.mean(swapped_units, axis=1), daily medians with np.median(swapped_units, axis=0), and population spread with np.std(swapped_units.reshape(-1)). Explain axes relative to the original units array. -->


```python

days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5', 'Day 6']
products = ['Notebook', 'Pen', 'Folder', 'Marker']

daily_totals = units.sum(axis=1)
product_means = units.mean(axis=0)
daily_medians = np.median(units, axis=1)
overall_std = np.std(units)

print('Total units per day:')
for i in range(6):
    print(days[i], ':', daily_totals[i])

print('Mean daily units per product:')
for i in range(4):
    print(products[i], ':', product_means[i])

print('Median across the four products, per day:')
for i in range(6):
    print(days[i], ':', daily_medians[i])

print('Overall population standard deviation:', overall_std)
```

**Explanation / findings:**

axis=0 works down the rows, so it gives a result for each product. axis=1 works across the columns, so it gives a result for each day. That’s why we use axis=1 for daily totals and axis=0 for product averages. The daily totals are 60, 57, 62, 68, 88, 72, and the product averages are 15.17, 31.5, 8.67, 12.5. The daily medians are 11, 13.5, 9.5, 16.5, 18, 15, and the population standard deviation is about 9.39.
